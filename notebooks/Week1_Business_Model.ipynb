## Business Understanding
"Customer churn is the rate at which customers stop doing business with a company. Losing a customer costs more than it seems, because replacing them requires marketing and sales expense, and the company also loses the future revenue that customer would have generated over time. The goal of the prediction model is to calculate a probability score showing how likely each customer is to churn, which gives more useful information than a simple yes/no answer. With this score, a company can group customers into risk tiers, and when a customer's churn probability crosses a high threshold (for example 70%), the company can proactively offer discounts or incentives to retain them, while lower-risk customers can simply be monitored."

## Data Quality Report

**Missing Values:** Found 11 rows with missing values in the TotalCharges column, where the missing values appeared as blank spaces (' ') rather than standard nulls, which is why they weren't caught by initial null checks. All 11 rows had a tenure of 0, meaning these were customers who had not yet completed a billing cycle, so no total charge had accumulated yet. All 11 were also non-churned customers. Handled by converting the column to numeric with pd.to_numeric(errors='coerce'), which converted the blanks to NaN, then filling those NaN values with 0, since a tenure of 0 mathematically corresponds to zero total charges billed.

**Duplicate Records:** Checked using df.duplicated() and found 0 duplicate rows across the full dataset. Also checked customerID specifically using df['customerID'].duplicated() and found 0 duplicates, confirming each customer appears exactly once.

**Data Type Issues:** The TotalCharges column was stored as text (object/str) instead of a numeric type, despite containing numeric values, because 11 rows contained a blank space instead of a number. Fixed by converting the column with pd.to_numeric(errors='coerce') followed by filling missing values with 0, resulting in a correct float64 dtype.

**Categorical Consistency:** Checked all 16 categorical columns for typos, inconsistent capitalization, and stray whitespace using .unique() on each column. Found no inconsistencies; all categorical values were clean and consistently formatted. Noted that several service-related columns (OnlineSecurity, OnlineBackup, DeviceProtection, TechSupport, StreamingTV, StreamingMovies) include a third value, "No internet service," which is a valid category rather than an error, since customers without internet service cannot have internet-dependent add-ons.
